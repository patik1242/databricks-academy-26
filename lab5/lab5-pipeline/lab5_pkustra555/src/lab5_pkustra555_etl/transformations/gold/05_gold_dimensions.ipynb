```python
from pyspark.sql import functions as F
from pyspark.sql import Window
from pyspark import pipelines as dp
```

```python

CATALOG = spark.conf.get("catalog")
SILVER_SCHEMA = spark.conf.get("silver_schema")
SILVER_TABLE = f"{CATALOG}.{SILVER_SCHEMA}.silver_netflix"

GOLD_SCHEMA = spark.conf.get("gold_schema")
```

```python
@dp.materialized_view(name=f"{GOLD_SCHEMA}.dim_type")
def dim_type():
    return(
        spark.read.table(SILVER_TABLE)
        .select("type")
        .distinct()
        .withColumn(
            "type_key", F.row_number().over(Window.orderBy("type"))
        )
        .select(
            "type_key",
            "type"
        )
    )
    
```

```python
@dp.materialized_view(name=f"{GOLD_SCHEMA}.dim_rating")
def dim_rating():
    return(
        spark.read.table(SILVER_TABLE)
        .select("rating", "audience_category")
        .distinct()
        .withColumn(
            "rating_key", F.row_number().over(Window.orderBy("rating"))
        )
        .select(
            "rating_key",
            "rating",
            "audience_category"
        )
    )
```

```python
@dp.materialized_view(name=f"{GOLD_SCHEMA}.dim_release_year")
def dim_release_year():
    return(
        spark.read.table(SILVER_TABLE)
        .select("release_year", "release_period")
        .distinct()
        .withColumn(
            "release_year_key", F.row_number().over(Window.orderBy("release_year"))
        )
        .select(
            "release_year_key",
            "release_year", 
            "release_period"
        )
    )
```
