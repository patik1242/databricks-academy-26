## Governance Test Tables

2. Row-Level Security

```python
%sql
CREATE OR REPLACE TABLE dbr_dev.pkustra555_gold.security_demo AS
SELECT
    show_id,
    title,
    type,
    rating,
    release_year
FROM dbr_dev.pkustra555_silver.silver_netflix;
```

```python
%sql 
CREATE OR REPLACE FUNCTION dbr_dev.pkustra555_gold.rls_filter(content_type STRING)
RETURN 
    CASE 
        WHEN current_user() LIKE '%softserve.academy' THEN TRUE 
        WHEN current_user() LIKE '%gmail.com' THEN content_type = 'TV Show'
        ELSE content_type = 'Movie'
    END; 
```

```python
%sql
ALTER TABLE dbr_dev.pkustra555_gold.security_demo
SET ROW FILTER dbr_dev.pkustra555_gold.rls_filter ON (type)
```

3. Column-Level Security

```python
%sql 
CREATE OR REPLACE FUNCTION dbr_dev.pkustra555_gold.rating_mask(rating STRING)
RETURN 
    CASE 
        WHEN current_user() LIKE '%softserve.academy' THEN rating
        ELSE 'RESTRICTED'
    END; 
```

```python
%sql 
ALTER TABLE dbr_dev.pkustra555_gold.security_demo
ALTER COLUMN rating 
SET MASK dbr_dev.pkustra555_gold.rating_mask
```
