```python
CATALOG = dbutils.widgets.get("catalog")
BRONZE_SCHEMA = dbutils.widgets.get("bronze_schema")
TITLE = dbutils.widgets.get("title")
spark.sql(
    f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{BRONZE_SCHEMA}"
)
spark.sql(
    f"""
    CREATE VOLUME IF NOT EXISTS {CATALOG}.{BRONZE_SCHEMA}.raw
    """
)
```

```python
BASE_PATH = f"/Volumes/{CATALOG}/{BRONZE_SCHEMA}/raw"

SOURCE_CSV = f"{BASE_PATH}/{TITLE}.csv"
PREPARED_PATH = f"{BASE_PATH}/prepared"
LANDING_PATH = f"{BASE_PATH}/landing"
REFERENCE_PATH = f"{BASE_PATH}/reference"

dbutils.fs.mkdirs(PREPARED_PATH)
dbutils.fs.mkdirs(LANDING_PATH)
dbutils.fs.mkdirs(REFERENCE_PATH)
```
