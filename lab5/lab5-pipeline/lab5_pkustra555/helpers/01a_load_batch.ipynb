```python
batch_number = int(dbutils.widgets.get("batch_id"))
CATALOG = dbutils.widgets.get("catalog")
BRONZE_SCHEMA = dbutils.widgets.get("bronze_schema")

BASE_PATH = f"/Volumes/{CATALOG}/{BRONZE_SCHEMA}/raw"

PREPARED_PATH = f"{BASE_PATH}/prepared"
LANDING_PATH = f"{BASE_PATH}/landing"
```

```python
source_dir = f"{PREPARED_PATH}/batch_{batch_number}"

csv_file = [
    file.path
    for file in dbutils.fs.ls(source_dir)
    if file.name.endswith(".csv")
][0]
```

```python
target_file = f"{LANDING_PATH}/batch_{batch_number}.csv"

dbutils.fs.cp(
    csv_file,
    target_file
)
```
