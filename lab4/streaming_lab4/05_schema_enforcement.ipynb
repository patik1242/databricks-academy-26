```python
from pyspark.sql import functions as F
from pyspark.sql import types as T

CATALOG = dbutils.widgets.get("catalog")
RAW_SCHEMA = dbutils.widgets.get("stream_schema")
TITLE = dbutils.widgets.get("title")
SILVER_SCHEMA = f"{RAW_SCHEMA}_silver"

SILVER_TABLE = f"{CATALOG}.{SILVER_SCHEMA}.{TITLE}"
```

```python
silver_df = spark.table(SILVER_TABLE)
silver_df.printSchema()
```

```python
bad_df = (
    silver_df.limit(1).withColumn("show_id", F.lit("s_bad_1")).withColumn("release_year", F.lit("not_a_year"))
    )

bad_df.printSchema()
```

```python
(
    bad_df.write
    .format("delta")
    .mode("append")
    .saveAsTable(SILVER_TABLE)    
)
```
