```python
dbutils.widgets.text("batch_number", "")
dbutils.widgets.text("catalog", "")
dbutils.widgets.text("stream_schema", "")

batch_number = dbutils.widgets.get("batch_number")
CATALOG = dbutils.widgets.get("catalog")
STREAM_SCHEMA = dbutils.widgets.get("stream_schema")

PREPARED_PATH = f"/Volumes/{CATALOG}/{STREAM_SCHEMA}/raw/{STREAM_SCHEMA}/prepared"
LANDING_PATH = f"/Volumes/{CATALOG}/{STREAM_SCHEMA}/raw/lab4_streaming/landing"
```

```python
display(dbutils.fs.ls(PREPARED_PATH))
```

```python
source_dir = f"{PREPARED_PATH}/batch_{batch_number}"

csv_file = [
    file.path
    for file in dbutils.fs.ls(source_dir)
    if file.name.endswith(".csv")
][0]
```

```python
target_file = f"{LANDING_PATH}/batch_{batch_number}.csv"

dbutils.fs.cp(
    csv_file,
    target_file
)
```
