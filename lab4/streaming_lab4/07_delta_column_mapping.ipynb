```python
from pyspark.sql import functions as F

CATALOG = dbutils.widgets.get("catalog")
RAW_SCHEMA = dbutils.widgets.get("stream_schema")
TITLE = dbutils.widgets.get("title")
SILVER_SCHEMA = f"{RAW_SCHEMA}_silver"

SILVER_TABLE = f"{CATALOG}.{SILVER_SCHEMA}.{TITLE}"
MAPPING_TABLE = f"{CATALOG}.{SILVER_SCHEMA}.column_mapping_test"
spark.sql(f"DROP TABLE IF EXISTS {MAPPING_TABLE}")
```

```python
spark.sql(f"""
CREATE TABLE {MAPPING_TABLE}
AS SELECT * FROM {SILVER_TABLE}
""")
```

```python
mapping_df = spark.table(MAPPING_TABLE)

mapping_df.printSchema()
```

```python
spark.sql(f"""
ALTER TABLE {MAPPING_TABLE} SET TBLPROPERTIES (
  'delta.columnMapping.mode' = 'name')
""")
```

```python
spark.sql(f"""
ALTER TABLE {MAPPING_TABLE} RENAME COLUMN rating TO age_rating
""")
```

```python
spark.sql(f"""
ALTER TABLE {MAPPING_TABLE} DROP COLUMN director
""")
```
